# Visualización documento

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import math
```

```python
traslados = pd.DataFrame({
    "año": [2019, 2020, 2021, 2022, 2023, 2024],
    "traslados": [339, 97, 146, 1104, None, None],
})
```

```python
traslados["variacion_pct"] = traslados["traslados"].pct_change(fill_method=None) * 100
```

```python
colores = []
for año, val in zip(traslados["año"], traslados["traslados"]):
    if val is None:
        colores.append("#D3D1C7")          # sin dato
    elif año == 2021:
        colores.append("#993C1D")          # renegociación de contratos
    elif año == 2022:
        colores.append("#185FA5")          # pico 
    else:
        colores.append("#B4B2A9")
```

```python
texto = [
    f"{int(v)}" if pd.notna(v) else "Sin dato"
    for v in traslados["traslados"]
]

fig = go.Figure(go.Bar(
    x=traslados["año"],
    y=traslados["traslados"],
    marker_color=colores,
    text=texto,
    textposition="outside",
    customdata=traslados["variacion_pct"].round(1),
    hovertemplate=(
        "<b>%{x}</b><br>"
        "Traslados: %{y}<br>"
        "Variación vs año anterior: %{customdata}%"
        "<extra></extra>"
    ),
))

fig.add_annotation(
    x=2022, y=1104,
    text="<b>+935% vs 2020</b>",
    showarrow=False,
    yshift=30,
    font=dict(size=13, color="#185FA5"),
    bgcolor="white", bordercolor="#185FA5", borderwidth=1,
)

fig.add_vline(
    x=2021, line_dash="dot", line_color="#993C1D", opacity=0.6,
    annotation_text="Renegociación de contratos",
    annotation_position="top left",
    annotation_font_size=11, annotation_font_color="#993C1D",
)

fig.update_layout(
    title=dict(
        text="Traslados de CERESOS a CEFERESOS (2019–2024)",
        font=dict(size=15), x=0.5, xanchor="center"
    ),
    xaxis=dict(title="Año", dtick=1),
    yaxis=dict(title="Número de traslados", range=[0, 1300]),
    plot_bgcolor="white",
    margin=dict(l=40, r=40, t=60, b=40),
    height=480,
    showlegend=False,
)

fig.show()
```

```python
indicadores = ["Calificación promedio (CNDH)", "Incidentes violentos", "Suicidios"]
ceferesos   = [8.39, 202, 47]
ceresos     = [6.55, 266, 102]
```

```python
color_cefereso = "#185FA5"
color_cereso   = "#993C1D"
```

```python
fig = make_subplots(
    rows=1, cols=3,
    subplot_titles=indicadores,
    horizontal_spacing=0.12,
)

for i, (label, val_cef, val_cer) in enumerate(zip(indicadores, ceferesos, ceresos), start=1):
    fig.add_trace(
        go.Bar(
            x=["CEFERESOS", "CERESOS"],
            y=[val_cef, val_cer],
            marker_color=[color_cefereso, color_cereso],
            text=[str(val_cef), str(val_cer)],
            textposition="outside",
            showlegend=False,
            hovertemplate="<b>%{x}</b><br>" + label.replace("<br>", " ") + ": %{y}<extra></extra>",
        ),
        row=1, col=i
        )

fig.add_trace(
    go.Bar(x=[None], y=[None], marker_color=color_cefereso, name="CEFERESOS", showlegend=True),
    row=1, col=1
)
fig.add_trace(
    go.Bar(x=[None], y=[None], marker_color=color_cereso, name="CERESOS", showlegend=True),
    row=1, col=1
)

fig.update_layout(
    title=dict(
        text="CEFERESOS vs CERESOS — indicadores CNDH 2024",
        font=dict(size=15), x=0.5, xanchor="center", y=0.97
    ),
    plot_bgcolor="white",
    margin=dict(l=40, r=40, t=100, b=40),
    height=480,
    legend=dict(orientation="h", yanchor="bottom", y=1.08, xanchor="center", x=0.5),
)

fig.update_yaxes(range=[0, 10],  row=1, col=1)
fig.update_yaxes(range=[0, 300], row=1, col=2)
fig.update_yaxes(range=[0, 120], row=1, col=3)

fig.show()
    
```

```python
actores_color = {
    "Carlos Slim": "#185FA5",
    "Olegario Vázquez Raña": "#3B6D11",
    "Fam. Salinas de Gortari": "#993C1D",
    "TRADECO / CROC": "#7A6C5D",
}

empresas = {
    "ICA": "Carlos Slim",
    "IDEAL": "Carlos Slim",
    "Capital Inbursa": "Carlos Slim",
    "PRODEMEX": "Olegario Vázquez Raña",
    "GIA": "Fam. Salinas de Gortari",
    "TRADECO/CROC": "TRADECO / CROC",
}

ceferesos = {
    "CEFERESO 11\n(Sonora)": "ICA",
    "CEFERESO 12\n(Guanajuato)": "ICA",
    "CEFERESO 13\n(Oaxaca)": "GIA",
    "CEFERESO 14\n(Durango)": "PRODEMEX",
    "CEFERESO 15\n(Chiapas)": "IDEAL",
    "CEFERESO 16\n(Morelos)": "Capital Inbursa",
    "CEFERESO 17\n(Jalisco)": "PRODEMEX",
    "CEFERESO 18\n(Coahuila)": "TRADECO/CROC",
}
```

```python
nodos = {}
for actor, color in actores_color.items():
    nodos[actor] = {"nivel": 0, "color": color}
for empresa, actor in empresas.items():
    nodos[empresa] = {"nivel": 1, "color": actores_color[actor]}
for cefereso, empresa in ceferesos.items():
    nodos[cefereso] = {"nivel": 2, "color": actores_color[empresas[empresa]]}

edges = []
for empresa, actor in empresas.items():
    edges.append((actor, empresa))
for cefereso, empresa in ceferesos.items():
    edges.append((empresa, cefereso))

niveles = {0: [], 1: [], 2: []}
for nombre, datos in nodos.items():
    niveles[datos["nivel"]].append(nombre)

# Ordenar nivel 2 agrupando por empresa, y nivel 1 agrupando por actor,
# para que cada "cuña" del círculo quede junta por color.
n0 = niveles[0]
n1 = sorted(niveles[1], key=lambda e: n0.index(empresas[e]))
n2 = sorted(niveles[2], key=lambda c: n1.index(ceferesos[c]))

radios = {0: 1.0, 1: 2.2, 2: 3.4}
pos = {}

def colocar_anillo(lista_nodos, radio, offset=0):
    n = len(lista_nodos)
    for i, nombre in enumerate(lista_nodos):
        angulo = 2 * math.pi * i / n + offset
        x = radio * math.cos(angulo)
        y = radio * math.sin(angulo)
        pos[nombre] = (x, y)

colocar_anillo(n0, radios[0])
colocar_anillo(n1, radios[1])
colocar_anillo(n2, radios[2])

edge_x, edge_y = [], []
for origen, destino in edges:
    x0, y0 = pos[origen]
    x1, y1 = pos[destino]
    edge_x += [x0, x1, None]
    edge_y += [y0, y1, None]

edge_trace = go.Scatter(
    x=edge_x, y=edge_y,
    mode="lines",
    line=dict(width=1, color="#D3D1C7"),
    hoverinfo="none",
    showlegend=False,
)

orden_nodos = n0 + n1 + n2
node_x = [pos[n][0] for n in orden_nodos]
node_y = [pos[n][1] for n in orden_nodos]
node_color = [nodos[n]["color"] for n in orden_nodos]
node_size = [36 if nodos[n]["nivel"] == 0 else 26 if nodos[n]["nivel"] == 1 else 18 for n in orden_nodos]

text_pos = []
for n in orden_nodos:
    x, y = pos[n]
    if nodos[n]["nivel"] == 0:
        text_pos.append("middle center")
    elif x >= 0:
        text_pos.append("middle right")
    else:
        text_pos.append("middle left")

node_trace = go.Scatter(
    x=node_x, y=node_y,
    mode="markers+text",
    marker=dict(size=node_size, color=node_color, line=dict(width=1.5, color="white")),
    text=orden_nodos,
    textposition=text_pos,
    textfont=dict(size=10),
    hoverinfo="text",
    showlegend=False,
)

fig = go.Figure(data=[edge_trace, node_trace])

fig.update_layout(
    title=dict(
        text="Red de actores privados en los CEFERESOS bajo esquema APP",
        font=dict(size=15), x=0.5, xanchor="center"
    ),
    xaxis=dict(visible=False, range=[-5, 5]),
    yaxis=dict(visible=False, range=[-4.5, 4.5], scaleanchor="x", scaleratio=1),
    plot_bgcolor="white",
    margin=dict(l=20, r=20, t=60, b=20),
    height=700,
    width=800,
)

fig.show()
```
